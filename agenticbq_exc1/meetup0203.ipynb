The Google Gen AI SDK lets you configure retry behavior through client parameters or **HttpRetryOptions**.

Key parameters

**initial_delay**: The initial delay in seconds before the first retry (Default: 1.0).

**attempts:** The maximum number of retry attempts (Default: 5).

**exp_base:** The base for the exponential backoff calculation (Default: 2).

**max_delay:** The maximum delay in seconds between retries (Default: 60).

**jitter:** A factor to add a random delay to the backoff (Default: 1).

**http_status_codes:** A list of status codes that trigger a retry.

```python
from google import genai
from google.genai import types
import base64
import os

def generate_prompt(prompt, model = "gemini-3.1-flash-lite-preview", image_uri = None):

  text1 = types.Part.from_text(text=prompt)

  parts = []
  parts.append(text1)
  print(parts)
    
  ref_image = None
  if image_uri:
    ref_image = types.Part.from_uri(
        file_uri=image_uri,
        mime_type="image/png",
    )
    parts.append(ref_image)


  contents = [
    types.Content(
      role="user",
      parts=parts
    )
  ]

  print(contents)

  generate_content_config = types.GenerateContentConfig(
    temperature = 1,
    top_p = 0.95,
    max_output_tokens = 32768,
    safety_settings = [types.SafetySetting(
      category="HARM_CATEGORY_HATE_SPEECH",
      threshold="OFF"
    ),types.SafetySetting(
      category="HARM_CATEGORY_DANGEROUS_CONTENT",
      threshold="OFF"
    ),types.SafetySetting(
      category="HARM_CATEGORY_SEXUALLY_EXPLICIT",
      threshold="OFF"
    ),types.SafetySetting(
      category="HARM_CATEGORY_HARASSMENT",
      threshold="OFF"
    )],
    thinking_config=types.ThinkingConfig(
      thinking_level="LOW",
    ),
  )

  resp = client.models.generate_content(
    model = model,
    contents = contents,
    config = generate_content_config,
    )

  return resp.candidates[0].content.parts[0].text
```

```python
from google import genai
from google.genai import types
import base64
import os

def generate_image(prompt, model = "gemini-3.1-flash-image-preview", reference_image_gcs_uri = None, reference_image_mome_type = "image/jpeg"):

  ##image1 = types.Part.from_uri(
  ##    file_uri="gs://cloud-samples-data/generative-ai/image/woman.jpg",
  ##    mime_type="image/jpeg",
  ##)
  ref_image = None
  if reference_image_gcs_uri:
    ref_image = types.Part.from_uri(
        file_uri=reference_image_gcs_uri,
        mime_type=reference_image_mome_type,
    )

  text1 = types.Part.from_text(text=prompt)

  parts = []
  if reference_image_gcs_uri:
    parts.append(ref_image)

  parts.append(text1)

  contents = [
    types.Content(
      role="user",
      parts=parts
    )
  ]

  generate_content_config = types.GenerateContentConfig(
    temperature = 1,
    top_p = 0.95,
    max_output_tokens = 32768,
    response_modalities = ["TEXT", "IMAGE"],
    safety_settings = [types.SafetySetting(
      category="HARM_CATEGORY_HATE_SPEECH",
      threshold="OFF"
    ),types.SafetySetting(
      category="HARM_CATEGORY_DANGEROUS_CONTENT",
      threshold="OFF"
    ),types.SafetySetting(
      category="HARM_CATEGORY_SEXUALLY_EXPLICIT",
      threshold="OFF"
    ),types.SafetySetting(
      category="HARM_CATEGORY_HARASSMENT",
      threshold="OFF"
    )],
    image_config=types.ImageConfig(
      aspect_ratio="9:16",
      image_size="1K",
      output_mime_type="image/png",
    ),
    thinking_config=types.ThinkingConfig(
      thinking_level="LOW",
    ),
  )

  resp = client.models.generate_content(
    model = model,
    contents = contents,
    config = generate_content_config,
    )

  ##print(resp)
  image = None
  for part in resp.parts:
    if part.text is not None:
        ##print(part.text)
        pass
    elif part.inline_data is not None:
        image = part.as_image()

  return image

```

```python
from google.cloud import storage
from google.cloud.exceptions import NotFound

def create_bucket_if_not_exists(bucket_name, project_id):
    """Creates a GCS bucket if it doesn't already exist."""
    storage_client = storage.Client(project=project_id)
    bucket = storage_client.bucket(bucket_name)

    try:
        storage_client.get_bucket(bucket)
        print(f"Bucket {bucket_name} already exists.")
    except NotFound:
        bucket = storage_client.create_bucket(bucket, location="us")
        print(f"Bucket {bucket_name} created.")
    return bucket
```

```python
def upload_file_to_bucket(bucket_name, source_file_name, project_id):
    """Uploads a file to the bucket."""
    storage_client = storage.Client(project=project_id)
    bucket = storage_client.bucket(bucket_name)
    blob = bucket.blob(source_file_name)

    blob.upload_from_filename(source_file_name)
    print(f"File {source_file_name} uploaded to {bucket_name}.")
```

```python
import base64
import requests
import google.auth
import google.auth.transport.requests

def generate_song(project_id, model_id, text_prompt, image_uri=None):
    """
    Calls the Vertex AI interaction endpoint with text and image inputs.
    """
    # Get credentials from the environment
    credentials, _ = google.auth.default(
        scopes=['https://www.googleapis.com/auth/cloud-platform']
    )
    auth_req = google.auth.transport.requests.Request()
    credentials.refresh(auth_req)

    url = f"https://aiplatform.googleapis.com/v1beta1/projects/{project_id}/locations/global/interactions"

    headers = {
        "Authorization": f"Bearer {credentials.token}",
        "Content-Type": "application/json"
    }

    # Construct the input parts
    input_parts = [{"type": "text", "text": text_prompt}]

    if image_uri:
        input_parts.append({
            "type": "image",
            "mime_type": "image/png",
            "uri": image_uri
        })


    payload = {
        "model": model_id,
        "input": input_parts
    }

    response = requests.post(url, headers=headers, json=payload)
    response.raise_for_status()
    return response.json()
```

```python
from google.cloud import storage
import io

def upload_bytes_to_gcs(project_id, bucket_name, destination_blob_name, byte_data, content_type = 'audio/mpeg'):
    try:
        # Initialize the GCS client
        storage_client = storage.Client(project=project_id)
        bucket = storage_client.bucket(bucket_name)
        
        blob = bucket.blob(f"{destination_blob_name}")
        blob.upload_from_string(byte_data, content_type=content_type)

        print(f"Success: Uploaded data to {bucket_name}/{destination_blob_name}")
        return True

    except Exception as e:
        print(f"Failed to upload: {e}")
        return False
```

```python
import time
import sys
from google import genai
from google.genai import types
from google.genai.types import GenerateVideosConfig, Image

def generate_video(prompt, output_gcs_uri, model = "veo-3.1-lite-generate-001",  image_uri =  None):

    operation = client.models.generate_videos(
        model=model,
        prompt=prompt,
        image=Image(
            gcs_uri=image_uri,
            mime_type="image/png",
        ),
        config=GenerateVideosConfig(
            aspect_ratio="16:9",
            generateAudio = True, 
            numberOfVideos = 1, 
            output_gcs_uri=output_gcs_uri,
        ),
    )

    while not operation.done:
        time.sleep(15)
        operation = client.operations.get(operation)
        print(operation)
    video_uri = None
    if operation.response:
        video_uri = operation.result.generated_videos[0].video.uri

    return video_uri
```

## Multimodal Gemini Prompts in BigQuery

We will build now multimodal prompt which will include refreence to image (poster), song (theme music), video (), speech (). Query needs to analyse if these things are consistent.

## Create table that aggregates all assets per event

```python
RAW_SQL=f"""
CREATE TABLE {PROJECT_ID}.{BIGQUERY_DATASET_NAME}.blended_assets AS 
WITH speech_data AS (
  SELECT 
    REGEXP_EXTRACT(ref.uri, r'_([^_]+)\.[^.]+$') AS city,
    ref AS speech_ref
  FROM `{PROJECT_ID}.{BIGQUERY_DATASET_NAME}.speeches`
),
image_data AS (
  SELECT 
    REGEXP_EXTRACT(ref.uri, r'_([^_]+)\.[^.]+$') AS city,
    ref AS image_ref
  FROM `{PROJECT_ID}.{BIGQUERY_DATASET_NAME}.posters`
),
video_data AS (
  SELECT 
    REGEXP_EXTRACT(ref.uri, r'gs://[^/]+/([^/]+)/') AS city,
    ref AS video_ref
  FROM `{PROJECT_ID}.{BIGQUERY_DATASET_NAME}.videos`
),
song_data AS (
  SELECT 
    REGEXP_EXTRACT(ref.uri, r'_([^_]+)\.[^.]+$') AS city,
    ref AS song_ref
  FROM `{PROJECT_ID}.{BIGQUERY_DATASET_NAME}.songs`
)
SELECT 
  COALESCE(s.city, i.city, v.city, sn.city) AS city,
  s.speech_ref,
  i.image_ref,
  v.video_ref,
  sn.song_ref
FROM image_data i
FULL OUTER JOIN speech_data s ON s.city = i.city
FULL OUTER JOIN video_data v ON COALESCE(s.city, i.city) = v.city
FULL OUTER JOIN song_data sn ON COALESCE(s.city, i.city, v.city) = sn.city
WHERE i.image_ref IS NOT NULL
"""

print(RAW_SQL)
```
